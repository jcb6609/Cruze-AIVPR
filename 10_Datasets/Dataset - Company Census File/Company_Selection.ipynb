```python
import pandas as pd # import pandas as module object pd
```

```python
"""
df = pd.read_csv('Company_Census_File_20260701.csv')
df.head()
"""
# To avoid a warning message for mixed data/variables inside Series obj (columns), let's use the next line of code:
df = pd.read_csv("Company_Census_File_20260701.csv", low_memory=False) # 'low_memory=False' tells Pandas: "Read the file first, then determine the data types."
df.head()
```

```python
df.info()
```

```python
# Sort columns alphabetically from A to Z
df = df.sort_index(axis=1) # 'axis=0' for rows, 'axis=1' for columns
```

```python
pd.set_option('display.max_columns', 147) # We could also use None as a second argument to display all columns even without knowing their total number, but since we know the total number of columns, then that would not be necessary
df.head(10)
```

```python
# What columns' info/names do we have?
pd.set_option('display.max_seq_items', None) # to display all sequential items (from for example 'df.columns'), None --> removes any display limit entirely
df.columns
```

```python
# Filter our data for only the state of Texas:
filt_df = (df[df['PHY_STATE'] == 'TX'])
len(filt_df) # the len(...) function calculates the number of rows in a ... df argument ('filt_df' in this case)
```

```python
# FLEETSIZE: (AI recommendation)
# O	56–75	✅ Include
# P	76–100	✅ Include
# Q	101–200	⭐ Include
# R	201–300	⭐ Include
# S	301–400	⭐ Include
# T	401–550	✅ Include

# Use the .isin() method with our refered filt_df Series obj 'FLEETSIZE' data/values (filt_df['FLEETSIZE']) with a list argument including all the information we want our new filtered to have, therefore avoiding using the or '|' operator many times
filt_df = (filt_df[filt_df["FLEETSIZE"].isin(["O", "P", "Q", "R", "S", "T"])])
len(filt_df)
```

```python
# filter to remove any rows with positive 'BUS_UNITS' column values whle keeping also NaN values (Not a Number)
# filt structure: df = (df[(df[]...) | (df[]...)])
filt_df = (filt_df[(filt_df['BUS_UNITS'] <= 0) | (filt_df['BUS_UNITS'].isna())]) # the method .isna() would include the NaN values four our df['Column'] reference data
len(filt_df)
```

```python
# POWER_UNITS = Straight trucks + truck tractors + buses
# Since Cruze focuses on Class 8 truck fleets (heaviest commercial vehicles) then we should be working on 'POWER_UNITS', which covers truck tractors data

# descending sort of 'POWER_UNITS' data/values
filt_df = filt_df.sort_values(by='POWER_UNITS', ascending=False) # 'ascending=True' --> ascending by default, 'ascending=False' --> descending
filt_df.head(30)
```

```python
filt_df['Match'] = (filt_df['POWER_UNITS'] == filt_df['TRUCK_UNITS']) # Check wether column values 'filt_df['POWER_UNITS']' and 'filt_df['TRUCK_UNITS']' 
pd.set_option('display.max_rows', None)
filt_df['Match'] # displays a boolean values in each row regarding if our previos Series obj data match
```

```python
# Count the number of unmatching data (False values) for our 'Match' column (filt_df['Match'])
count = 0
for i in filt_df['Match']:
    if (i == False):
        count += 1
print(count)

# Out of 695 values, there are 87 rows data regarding 'POWER_UNITS' and 'TRUCK_UNITS' columns that do not match, approx a 12% of our total (0.124 * 695 = 86.18)
# 'Losing more than 2% to 5% of your dataset without investigation can introduce severe bias into your final analysis'.
# Then, the idea of merging columns is discarded, and rather keep working with only with the 'POWER_UNIT' Series obj
```

```python
# Creating a 'Qualified_Prospects.csv' out of our filt_df and valuable columns:
prospect_cols = [
    "LEGAL_NAME",
    "DOT_NUMBER",
    "POWER_UNITS",
    "FLEETSIZE",
    "TOTAL_DRIVERS",
    "PHY_CITY",
    "PHY_STATE",
    "PHY_STREET",
    "PHY_ZIP",
    "PHONE",
    "EMAIL_ADDRESS"
]

qualified_prospects = (filt_df[prospect_cols]) # Save filtered data with another name ('qualified_prospects'), and based on column-specific Series objs 'prospect_cols' info
qualified_prospects.to_csv("Qualified_Prospects.csv", index=False) # .to_csv() method creates a new csv out of the referenced df, and the second argument 'index=False' basically avoids passing an index column to our csv creation
```
